# **Project 6: Countdown Timer Python Project**

```python
# Project 5 " Count Down Timer In Python.
# Description: Ek count down timer he jo user s seconds me time input leta he aur phir usko
# minute second format main displaye krta he.


import time
def countdown_timer(seconds):
    while seconds > 0:
        mins, secs = divmod(seconds, 60)  #minutes aur seconds calculate horhe hen.
        time_format = '{:02d}:{:02d}'.format(mins, secs) #MM:SS format main time displaye ho raha he.
        print(time_format, end='\r')
        time.sleep(1)
        seconds -= 1
print("00:00 \n Time's Up!")

# user input for time .
total_seconds = int(input("Enter time in second for countdown: "))
countdown_timer(total_seconds)


```
